```python
import easyocr
import cv2
import matplotlib.pyplot as plt
from google.colab import files
```

```python
uploaded = files.upload()
```

```python
reader = easyocr.Reader(['en'])
```

```python
for filename in uploaded.keys():
    img = cv2.imread(filename)
    results = reader.readtext(img)

    print(f"\n--- {filename} ---")
    for (bbox, text, confidence) in results:
        print(f"Text: {text} | Confidence: {confidence:.2f}")
```

```python
import re

def extract_plate_candidate(results):
    candidates = []
    noise_words = ['IND', 'INDIA', 'INDI', 'TOURIST', 'PERMIT', 'ALL', 'NDIA']
    for (bbox, text, conf) in results:
        cleaned = re.sub(r'[^A-Z0-9]', '', text.upper())
        if cleaned in noise_words:
            continue
        if 7 <= len(cleaned) <= 12 and any(c.isdigit() for c in cleaned) and any(c.isalpha() for c in cleaned):
            candidates.append((cleaned, conf))
    if not candidates:
        return None
    candidates.sort(key=lambda x: x[1], reverse=True)
    return candidates[0]

for filename in uploaded.keys():
    img = cv2.imread(filename)
    results = reader.readtext(img)
    best = extract_plate_candidate(results)
    print(f"{filename}: {best}")
```

```python
def find_plate_candidates(img, top_n=3):
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    gray = cv2.bilateralFilter(gray, 11, 17, 17)
    edged = cv2.Canny(gray, 30, 200)
    contours, _ = cv2.findContours(edged.copy(), cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)
    contours = sorted(contours, key=cv2.contourArea, reverse=True)[:30]
    candidates = []
    for c in contours:
        x, y, w, h = cv2.boundingRect(c)
        aspect = w / float(h) if h > 0 else 0
        area = w * h
        if 2.0 <= aspect <= 6.0 and area > 1000:
            candidates.append((x, y, w, h))
    return candidates[:top_n]

for filename in uploaded.keys():
    img = cv2.imread(filename)
    candidates = find_plate_candidates(img)
    print(f"\n--- {filename} ---")
    if not candidates:
        print("No plate-shaped region found")
        continue
    for i, (x, y, w, h) in enumerate(candidates):
        crop = img[y:y+h, x:x+w]
        results = reader.readtext(crop)
        text = " ".join([r[1] for r in results])
        conf = max([r[2] for r in results], default=0)
        print(f"Candidate {i+1} (box {w}x{h}): '{text}' | max conf: {conf:.2f}")
```

```python
import re

def clean_text(text):
    text = text.upper()
    text = re.sub(r'\bIND[I]?A?\b', '', text)  # IND/INDI/INDIA hatao
    cleaned = re.sub(r'[^A-Z0-9]', '', text)   # sirf letters+numbers rakho
    return cleaned

def get_final_plate(img):
    candidates = find_plate_candidates(img)
    best_result = None
    best_conf = 0
    for (x, y, w, h) in candidates:
        crop = img[y:y+h, x:x+w]
        results = reader.readtext(crop)
        combined_text = " ".join([r[1] for r in results])
        cleaned = clean_text(combined_text)
        conf = max([r[2] for r in results], default=0)
        if 7 <= len(cleaned) <= 12 and conf > best_conf:
            best_result = cleaned
            best_conf = conf
    return best_result, best_conf

for filename in uploaded.keys():
    img = cv2.imread(filename)
    plate, conf = get_final_plate(img)
    print(f"{filename}: {plate} (confidence: {conf:.2f})")
```
